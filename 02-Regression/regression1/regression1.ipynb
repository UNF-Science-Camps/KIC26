### Beregning af fejlagtighed
Nu hvor vi har en meget simpel måde at bestemme en tendens linje, kunne det være dejligt at kunne måle hvor god tendensen til at repræsentere dataet- vi har brug for en måde at måle hvor akkurat den er.
Det kan vi gøre ved brug af **SSE** (*sum of squared errors*) metoden, som illustreret ovenover; afvigelserne mellem tendensen og dataet kvadrateres summeres sammen.

For hvert punkt kan vi måle hvor meget vores model rammer ved siden af:
$$\text{fejl} = \text{model}(x) - y$$
Vi kvadrerer fejlen (Så plus og minus ikke kan udligne hinanden, dette sørge også for at store fejl straffes hårdere end små) og lægger dem sammen over alle punkter:
$$SSE = \sum (y-\text{model}(y))^2 = (\text{fejl}_1)^2 + (\text{fejl}_2)^2 + ... \qquad \text{jo lavere, jo bedre fit}$$

**Perspektivering:** 

Fra *Excel* kender i nok $R^2$ fra "tilføj tendenslinje" — tættere på 1 er bedre fit. $R^2$ bygger faktisk på SSE:
$$R^2 = 1 - \frac{SSE}{SST}$$
$$\text{hvor } SST = \sum (y-\bar y)^2 = \text{ SSE hvis modellen altid gætter gennemsnittet af } y$$

**MSE:**: 

I praksis anvendes  MSE (*mean squared error*) metoden over SSE idet den angiver den gennemsnitlige afvigelsen, og derfor er nemmere at bruge til sammenligning.
MSE er lig gennemsnittet af de kvadrerede fejl, altså SSE, delt med antal punkter $n$:
$$MSE = \frac{SSE}{n}$$
Dette sørge for at fejlen er sammenlignelig på tværs af datasæt med forskelligt antal punkter, så er størrlsesordnen af fejl sammenlignelig hvis man har 3 punkter eller 3000 punkter

### Loss minimering via. differentialregning
Vi ønsker at finde den retning vi kan flytte parameterne for at bevæge sig mod lavere loss.

Hvis vi matematisk kunne bestemme hvordan loss ville blive påvirket hvis vi skubbede parameter i en hvis retning, kan man forestille sig at kunne forbedre vores loss iterativt ved blot at skubbe vores parametre mod den retning der forbedrer loss mest, gentaget indtil vi er tilfredse.

Dette kan vi gøre ved at bestemme hældningen på vores loss funktion (tænk igen på loss som en overflade). Dette kan gøres matematisk via differentiering.
Det er derfor en prerequsite at have kendskab til at differentiere (ca. 2g-3g pensum i gym). At differentiere er konceptuelt at finde hældningen af en graf i et punkt.

Differentiering er ikke altid lige nemt, men med kun 3 regler, kan man bygge sig en rimelig omfattende intuition for metoden- vær dog ops på at der findes mange flere regler.
Husk at det ikke nødvendigvis er vigtigt at du har en fuldstændig evne til at differentiere i alt det her- det vigtige her er mest bare at du for en ide om hvordan at man kan bruge differentiering til at finde hældningen af funktioner matematisk. Det er værktøjet til at nå endemålet, ikke selve endemålet.

**potensreglen:**
$$f(x) = c\cdot x^n \quad\Rightarrow\quad \frac{df}{dx}(x) = f'(x) = c\cdot n\cdot x^{n-1}$$
eg:
$$f(x) = c\cdot x^2 \quad\Rightarrow\quad \frac{df}{dx}(x) = f'(x) = c\cdot 2\cdot x$$

**sumreglen** — differentiér hvert led for sig, og læg sammen:
$$f(x) = g(x) + h(x) \quad\Rightarrow\quad \frac{df}{dx}(x) = f'(x) = g'(x) + h'(x)$$

**kædereglen** — når en funktion sidder inden i en anden (her: noget i anden), differentiér det ydre, og gang med den afledte af det indre:
$$f(x) = g(h(x)) \quad\Rightarrow\quad \frac{df}{dx}(x) = f'(x) = g'(h(x))\cdot h'(x)$$
eg:
$$f(x) = h(x)^2 \quad\Rightarrow\quad \frac{df}{dx}(x) = f'(x) = 2\cdot(h(x))\cdot h'(x)$$

### Differentiering ift. flere variable
De tidligere funktioner har kun haft én variabel at differentiere ift. til, $x$. Men loss-funktionen fra opgave 3 afhænger jo af flere parametre ($a$, $b$, evt. $c$) på samme tid. Derfor skal vi kunne differentiere funktioner af flere variabler.

Det gør vi med **partielt afledte**: vi differentierer mht. én variabel ad gangen, og behandler alting som ikke er direkte påvirket af variablet som om de er konstanter.
For en funktion $f(x,y)$ skriver vi det $\dfrac{\partial f}{\partial x}$ og $\dfrac{\partial f}{\partial y}$ både potens-, sum-, og kæderegel fra før kan anvendes inden for en enkelt partielt afledte af gangen.

Term: **gradient af $f$** = en vektor der indeholder alle de partielt afledte i et punkt:
$$\nabla f = \left(\frac{\partial f}{\partial x}, \frac{\partial f}{\partial y}\right)$$
Den peger i den retning $f = loss$ vokser hurtigst.
Gradient er et vigtigt redskab når vi ønsker at bevæge os ned af loss-fladen mod en model med et bedre fit og lavere loss.
Tænk på gradienten som at angive den retning du kan skubbe alle parametrene minimalt på en gang for at forbedre loss funktionen.
den angiver altså ikke det præcise sted for at loss er lavest- bare hvordan du kan skubbe parametrene så loss bliver mindre.
Det betyder i praksis også at selv hvis du har parametre som ikke kan nudges uden at gøre loss værre, er dine parametre ikke nødvendigvis optimalt indstilte, da det kan tænkes at en anden kombination af parametre stadig kan få en bedre loss.

### Gradienter ift. ML

Lad os nu samle det vi har lært om **modellen**, **fejlen** og **SSE**:

*syntax note: jeg bruger subscript $a,b$ for at highlighte at det er beregnet i et punkt på fladen*

$$\text{model}_{a,b}(x) = a\cdot x + b$$
$$\text{fejl}_{a,b}(x,y) = \text{model}_{a,b}(x) - y$$
$$SSE_{a,b}(\text{punkter}) = \text{fejl}_{a,b}(x_1,y_1)^2 + \text{fejl}_{a,b}(x_2,y_2)^2 + \dots$$

Vi ønsker at finde gradienten af SSE fladen, og hældningen i punktet $(a,b)$:
$$\nabla SSE_{a,b} = \left(\frac{\partial SSE_{a,b}}{\partial a}, \frac{\partial SSE_{a,b}}{\partial b}\right)$$

Det gør vi ved at kombinere kædereglen med partielt afledte, SSE er jo summen af kvadrater af nogen ($\text{fejl}$), der selv afhænger af modellens parametre.

**Bemærk:**

Før beregnede vi de afledte med respekt til x og y, nu beregner vi de afledte med respekt til a og b.

**Hint:**

Per **sumreglen** kan vi i vores beregning af $\nabla SSE_{a,b}$ beregne hvert punkt for sig før de til sidst lægges sammen.

### Opgave 8)
Fordi i har regnet gradienten af SSE ud (opgave 7), er det let at finde gradienten af MSE også.
MSE er SSE delt med en konstant $n$ (antal punkter), når f er ganget med en konstant ganges gradienten af f også med den konstant:
$$MSE = \frac{SSE}{n} \quad\Rightarrow\quad \nabla MSE = \frac{\nabla SSE}{n}$$

MSE er faktisk det man oftere ser blive brugt i praksis til gradient descent.
Med SSE bliver gradienten (og dermed det gode valg af learning rate) afhængig af hvor mange datapunkter man har; med MSE behøver man ikke tage højde for antallet af punkter.

### Gradient descent
Gradienten peger i den retning loss vokser hurtigst (som pilen tideligere viste), så for at gå NED ad fladen skal vi bevæge os den modsatte retning:

Vi har et sæt start parametre, og så går vi et lille skridt imod gradienten, for at få en lavere loss, det er hele ideen i gradient descent.

For at få en intuition for hvorledes gradient decent fungere generelt, så starter vi med en simpel model ($y=a\cdot x+b$).
Gradient Decent kan brydes ned i følgende del processer:
1. **opdater vægtene** — I har allerede skridtet (retningen for at gøre dette), træk det bare fra (a_delta/b_delta) (vores nuværende parameter værdier)
2. **ét helt gradient descent-skridt** — brug jeres egen opg8_1 (MSE-gradienten fra opgave 8) til selv at beregne skridtet, og jeres opdatering fra 1. til at udføre det
3. **et træningsloop** — gentag skridtet fra 2. n gange

De senere opgaver til sidst gør vi det samme for parabel modellen: $y=a\cdot x^2+b\cdot x+c$, med opg8_2 i stedet.

### Bonus indhold
**cool: closed form solution** 

Til de simple modeller vi har set frem til nu findes der "closed form solutions" men det gør der ikke når vi laver neurale netværk

closed-form-løsningen er netop det sted hvor gradienten af loss er $0$ 

Sådanne løsninger findes for problemer på formen:

$$loss = y = a\cdot f(x) + b\cdot g(x) + ...$$

hvor $a, b, ...$ kan ændres vægt på, men $f(x), g(x), ...$ forskrifterne ikke kan modificeres, og man ønsker at finde de $a, b, ...$ som minimere $loss = y$

```python
# leg med at sætte punkter og se den bedste rette linje visualiseret
punkter = [(1,2),(3,4),(4,3),(2.5,2.5)]  # prøv at tilføje eller ændre punkterne

xs = np.array([p[0] for p in punkter])
ys = np.array([p[1] for p in punkter])
A  = np.column_stack([xs, np.ones_like(xs)])   # hver række: [x, 1]

AtA     = A.T @ A
Aty     = A.T @ ys
AtA_inv = np.linalg.inv(AtA)   # den eneste "sorte boks" her — resten er bare matrix-multiplikation
w       = AtA_inv @ Aty
a, b    = w
print(f"a = {a:.4f},  b = {b:.4f}")

loss_fn = lambda a, b: opg3_2(a, b, punkter)
grad_a, grad_b = opg7_3(a, b, punkter)   # jeres egen gradient-funktion fra opgave 7
print(f"gradient ved (a,b) = ({grad_a:.2e}, {grad_b:.2e})")   # ~0 — derfor ingen pil i konturplottet

sesy_viz.display(
    sesy_viz.linalg_losning(A, ys, AtA, Aty, AtA_inv, w),
    sesy_viz.loss_kontur(loss_fn, a_range=(-2,3), b_range=(-2,3), punkt=(a,b), gradient=(grad_a, grad_b)),
    sesy_viz.modelfit(a, b, punkter),
)

# jeres eget træningsloop (opgave 9) fra nul, mange nok skridt til at det når helt derned:
a_gd, b_gd = opg9_3(0.0, 0.0, punkter, lr=0.1, n=500)
print(f"optimalt punkt for linjefitting igennem punkterne: a = {a_gd:.4f},  b = {b_gd:.4f}")
```
