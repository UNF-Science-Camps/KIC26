## Hvorfor vi splitter vores data
Til at starte med, lad os glemme om valideringssættet og istedet fokusere kun på vores fulde dataset. Hvis du træner en model til at genkende mønstre i din data kan den typisk finde en løsning ret hurtigt og ret præcist med nok kompleksitet og lidt træning. 
<br><br>
Dette lyder godt, men det er virkelig ikke særlig imponerende fordi det ville svarer til at du giver en klog model din prøve og svar-arket og beder den om at finde svarene. Hurtigt ville din model finde ud af at den kan ignorere spørgsmålene helt og i stedet fokusere på kun at huske svar-arket, da det er det eneste du er interesseret i.
<br><br>
I stedet ville det være bedre hvis du baserede din models præstation på en ny prøve på samme emne, men med andre spørgsmål som den ikke har set før.
<br><br>
Når vi splitter vores data for første gang er dette hvad vi gør. Og på den måde stopper vi vores <strong>model</strong> fra at overfitte.
</div><br>

```python
# Eksempel 4.1: Sådan splitter du din data
data = list(range(10_000))
```

Vi har indlæst vores 10,000 "datapunkter" i en liste. Nu vil vi så splitte dem op i et trænings- og testsæt.

```python
# manuel implementeringt
def split_data(data, test_size=0.2):
    n = len(data)

    # find split at 1 - test-size
    n_split = int((1-test_size) * n)

    train = data[:n_split]
    test = data[n_split:]

    return train, test

train, test = split_data(data, test_size=0.2)

print(f'Train points:\t{len(train)}\t{train[:5]}')
print(f'Test points:\t{len(test)}\t{test[:5]}')
```

Heldigvis kan vi se at vores metode splitter vores data punkter ind i et træningssæt af 0.80 \* 10000 = 8000 punkter og 0.20 \* 10000 = 2000 punkter, ergo har vi nu splittet vores data som man ville gøre i praksis.
<br><br>
Men nu hvor vi har set hvordan man implementerer det vil har set hvordan det implementeres vil vi aldrig implementere det igen og i stedet lade scikit-learn klare alt det hårde arbejde for os. Scikit-learns split-funktion er også bedre da du selv kan vælge om du vil blande din data eller stratificere den.


```python
from sklearn.model_selection import train_test_split
train, test = train_test_split(data, test_size=0.2, random_state=1000)

print(f'Train points:\t{len(train)}\t{train[:5]}')
print(f'Test points:\t{len(test)}\t{test[:5]}')
```

Læg mærke til indekserne af vores train- og test split. I eksemplet ovenfor kan vi se at de første fem træningspunkter læser 0,1,2,3,4 og at vores testpunkter læser 8000,8001,8002,8003,8004 som et resultat af at vi bare lagde alle vores punkter op på en lige linje og klippede den over ved 80% markøren. Kigger vi i stedet på vores scikit-learn split kan vi se at alle vores punkter virker helt tilfældige. Dette er med vilje, og er en konsekvens af at scikit-learn's train-test-split funktion automatisk har shuffle=True til.

## Hvorfor splitter vi vores data... igen?
Som vi så ovenfor splitter vi vores data første gang for at fjerne bias i vores model, så vores resultater er mere troværdige og i en højere grad reflektere sand læring. I mange kurser hører man skræmme historier om forskere som spændt deler deres virkelig imponerende resultater på forsiden, kun til at blive pinliggjort da det viser sig de ikke havde splittet deres data og dermed kun opnået lav training loss (overfitting). Med det sagt vidste de forskere nok ikke bedre, og ligevis nutildags, hvis man kun splitter sin data én gang er man næsten ligeså cooked.

Grunden til vi splitter igen opstår i det vi inddrager ***hyperparametre***. En hyperparameter er en parameter som ikke direkte påvirkes af gradienten når din model lærer. I stedet er hyperparametre typisk valg du selv tager i forhold til din models træning – nogen gange i formen af en uformel beslutning, andre gange i formen af et tal repræsenteret af et matematisk symbol (nøjagtigt ligesom en parameter 😭).

Når vi splitter vores data igen så vi har et træningsset, testset, og et ***valideringsset***, kan vi optimere vores **parametre** igennem træningssettet, optimere vores **hyperparametre** igennem valideringssettet, og til sidst **teste/evaluere** vores model igennem testsettet.
<br><br>

```python
# eksempel 4.2: sådan splitter du din data... igen
data = list(range(10_000)) # list fra 0...9999
```

# 6: Bias-Variance Tradeoff
Lidt i samme tråd som med model kompleksitet, siger vi der er en balance man skal overveje når man vælger den rette model for sit problem. I denne sektion vil du fokusere på det vi kalder bias-varians balancen – et fænomen som opstår når man ser ens modeller som hypoteser hvori man må lave antagelser.

Realiteten er at hovedopgaven i ML kan ses fra flere forskellige _paradigmer_, hvor, afhængigt af hvilket paradigme man ser sit problem ud fra, man ender med at tilgå sit problem på forskellige måder. Hvilket paradigme som er det korrekte for hvilket problem ved ingen, så hvordan man vælger det kommer meget an på baggrund, erfaring og personlighed som data scientist.
<br>

## Modeller som Hypoteser
Én vigtig måde at forestille sig ML problemet er som at lave en række uddannede gæt (antagelser) efterfulgt af en løsning inden for gættenes rammer. Antagelser er gode fordi de hjælper os gøre vores problem "lettere" idet vi begrænser kompleksiteten af den løsning vi tror karakteriserer problemet.

Et godt eksempel på en antagelse som gør et problem meget lettere at løse er hvis man forestiller sig at Nord Korea går i krig mod Danmark fordi de gerne vil have Grønland. Kong Frederik har bedt dig om at lave en model som kan forudse om Danmark ville vinde i en krig mod Nord Korea, fordi hvis ikke, så kunne Danmark bare give op tidligt og spare en masse liv og velfærd ved at undgå krigen til at starte med.

Hvis vi var meget naive kunne vi lave den smarte antagelse at fodbold evner er fuldstændigt korrolerede med krigsførselsevner – altså hvem end der er bedst til fodbold ville vinde krigen. Problem løst!! Danmark er ret gode til fodbold og Nord Korea er desværre ikke ligeså gode, derfor ville Danmark vinde krigen ret nemt.

Som det viser sig blev vores komplekse problem faktisk helt trivielt idet vi lavede en meget biased antagelse. Derudover varierer resultatet vi fik fra denne antagelse ikke så meget baseret på de variable vi ellers typisk kigger på når man diskutterer krig, og dermed ville resultaterne have lav varians.

Helt præcist: Højt bias lod til lav varians.

Vender vi dette fænomen om og laver en model som kigger på hver lille detalje af krigsførsel vil vores problem begynde at blive mere sensitivt til små dog relevante detaljer (høj varians), netop som den pris vi betaler for at have små, realistiske antagelser (lavt bias).

Lavt bias lod til høj varians.
